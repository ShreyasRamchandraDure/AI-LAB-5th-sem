```python

# First Order Logic Resolution

# Knowledge Base (Facts)
facts = [
    "Alive(Anil)",
    "Eats(Anil,Peanuts)"
]

# Query to prove
query = "Likes(John,Peanuts)"

print("KNOWLEDGE BASE")
print("---------------------------")

for fact in facts:
    print(fact)

print("\nRULES")
print("---------------------------")
print("1. Alive(x) -> NotKilled(x)")
print("2. Eats(x,y) AND NotKilled(x) -> Food(y)")
print("3. Food(x) -> Likes(John,x)")

print("\nRESOLUTION STEPS")
print("---------------------------")

# Step 1: Anil is alive
print("Step 1: Alive(Anil)")

# Step 2: Alive implies not killed
not_killed = True
print("Step 2: NotKilled(Anil)")

# Step 3: Anil eats peanuts and is not killed
if "Eats(Anil,Peanuts)" in facts and not_killed:
    food = True
    print("Step 3: Food(Peanuts)")

# Step 4: John likes all food
if food:
    likes = True
    print("Step 4: Likes(John,Peanuts)")

# Step 5: Negate the query
print("Step 5: Negated query = Not Likes(John,Peanuts)")

# Step 6: Resolution
if likes:
    print("Step 6: Empty Clause (contradiction)")

    print("\nRESULT")
    print("---------------------------")
    print("Query proved successfully!")
    print("John likes peanuts.")
else:
    print("Query cannot be proved.")
```
