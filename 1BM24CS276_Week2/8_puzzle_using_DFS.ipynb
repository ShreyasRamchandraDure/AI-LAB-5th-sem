```python
def dfs(start, goal, max_depth=30):
    stack = [(start, [start])]

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path

        if len(path) > max_depth:
            continue

        zero = state.index(0)
        row = zero // 3
        col = zero % 3

        moves = []

        if row > 0:
            moves.append(zero - 3)   # Up
        if row < 2:
            moves.append(zero + 3)   # Down
        if col > 0:
            moves.append(zero - 1)   # Left
        if col < 2:
            moves.append(zero + 1)   # Right

        for move in moves:
            new_state = list(state)

            new_state[zero], new_state[move] = \
                new_state[move], new_state[zero]

            new_state = tuple(new_state)

            if new_state not in path:
                stack.append((new_state, path + [new_state]))

    return None


def display(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


start = (5, 4, 6,
         0, 1, 8,
         7, 3, 2)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = dfs(start, goal, 30)

if solution:
    print("Solution found:")
    print("Number of moves:", len(solution) - 1)

    for state in solution:
        display(state)
else:
    print("No solution found within depth limit.")
```
