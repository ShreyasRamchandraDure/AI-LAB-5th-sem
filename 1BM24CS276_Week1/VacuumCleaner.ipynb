```python
# Two Room Vacuum Cleaner Problem

room_a = input("Enter status of Room A: ")
room_b = input("Enter status of Room B: ")
position = input("Enter vacuum position (A/B): ")

if position == "A":
    if room_a == "Dirty":
        print("Room A is Dirty -> Cleaning Room A")
        room_a = "Clean"

    print("Moving to Room B")

    if room_b == "Dirty":
        print("Room B is Dirty -> Cleaning Room B")
        room_b = "Clean"

else:
    if room_b == "Dirty":
        print("Room B is Dirty -> Cleaning Room B")
        room_b = "Clean"

    print("Moving to Room A")

    if room_a == "Dirty":
        print("Room A is Dirty -> Cleaning Room A")
        room_a = "Clean"

if room_a == "Clean" and room_b == "Clean":
    print("Both rooms are clean.")
    print("Goal Reached!")
```
