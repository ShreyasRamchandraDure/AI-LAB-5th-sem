```python

# Forward Chaining in First Order Logic

facts = []

n = int(input("Enter number of facts: "))

for i in range(n):
    fact = input("Enter fact (Predicate,Value1,Value2): ")
    parts = fact.split(",")
    facts.append(tuple(x.strip() for x in parts))

print("\nEnter Query:")
query = input("Enter query (Predicate,Value1,Value2): ")
query = tuple(x.strip() for x in query.split(","))

print("\nInitial Facts:")
for f in facts:
    print(f"{f[0]}({f[1]}, {f[2]})")

print("\nApplying Forward Chaining...\n")

# Rule:
# Parent(X,Y) AND Parent(Y,Z) -> Grandparent(X,Z)

new_facts = []

for f1 in facts:
    for f2 in facts:
        if (f1[0] == "Parent" and
            f2[0] == "Parent" and
            f1[2] == f2[1]):

            new_fact = ("Grandparent", f1[1], f2[2])

            if new_fact not in facts and new_fact not in new_facts:
                new_facts.append(new_fact)
                print("New Fact Generated:")
                print(f"{new_fact[0]}({new_fact[1]}, {new_fact[2]})")

facts.extend(new_facts)

print("\nChecking Query:")
print(f"{query[0]}({query[1]}, {query[2]})")

if query in facts:
    print("\nQuery Proved Successfully!")
    print("Result: TRUE")
else:
    print("\nQuery Not Proved!")
    print("Result: FALSE")
```
