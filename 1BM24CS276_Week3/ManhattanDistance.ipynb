```python
import heapq

def get_state():
    print("Enter 9 values (use 0 for blank):")
    return tuple(map(int, input().split()))

initial = get_state()

print("Enter goal state:")
goal = get_state()


def manhattan_distance(state):
    distance = 0

    for i in range(9):

        if state[i] == 0:
            continue

        current_row = i // 3
        current_col = i % 3

        goal_position = goal.index(state[i])

        goal_row = goal_position // 3
        goal_col = goal_position % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def a_star():

    open_list = []

    g = 0
    h = manhattan_distance(initial)
    f = g + h

    heapq.heappush(open_list, (f, g, initial))

    parent = {initial: None}
    cost = {initial: 0}

    while open_list:

        f, g, current = heapq.heappop(open_list)

        if current == goal:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\nSolution Path:")

            for state in path:
                print_state(state)

            print("Number of moves:", len(path) - 1)

            return

        for next_state in get_neighbors(current):

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g

                h = manhattan_distance(next_state)
                f = new_g + h

                heapq.heappush(
                    open_list,
                    (f, new_g, next_state)
                )

                parent[next_state] = current

    print("No solution found")


a_star()
```
