```python

def unify(x, y, substitution):

    if x == y:
        return True

    # If x is a variable
    if x.islower() and len(x) == 1:
        if x in substitution:
            return unify(substitution[x], y, substitution)
        substitution[x] = y
        return True

    # If y is a variable
    if y.islower() and len(y) == 1:
        if y in substitution:
            return unify(x, substitution[y], substitution)
        substitution[y] = x
        return True

    return False


# Input
expr1 = input("Enter first expression: ")
expr2 = input("Enter second expression: ")

def extract(expr):
    name, args = expr.split("(")
    args = args.replace(")", "").split(",")
    return name.strip(), [a.strip() for a in args]


p1, a1 = extract(expr1)
p2, a2 = extract(expr2)

substitution = {}

# Check predicate and arguments
if p1 != p2 or len(a1) != len(a2):
    print("\nUnification Failed")

else:
    success = True

    for x, y in zip(a1, a2):
        if not unify(x, y, substitution):
            success = False
            break

    if success:
        print("\nUnification Successful")
        print("Substitution:")

        for key, value in substitution.items():
            print(key, "=", value)

        result = [substitution.get(x, x) for x in a1]

        print("\nUnified Expression:")
        print(p1 + "(" + ", ".join(result) + ")")

    else:
        print("\nUnification Failed")
```
